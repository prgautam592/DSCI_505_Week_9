```python
# sqlite3 provides tools for connecting to SQLite databases.
import sqlite3
# pandas works with tables; pd is its short name.
import pandas as pd
# Connect to the database (creates the file if it doesn't exist)
conn = sqlite3.connect('my_data.db')
# Create a cursor object to execute SQL commands
cursor = conn.cursor()
```

# 1. SELECT: Sales Data
Display sale_date, product, and price from the sales table.

```python
# Load sales.csv into the 'sales' table
pd.read_csv('sales.csv').to_sql('sales', conn, if_exists='replace', index=False)
# Preview 5 rows to see all the columns in the sales table
pd.read_sql_query('''
SELECT * FROM sales LIMIT 5
''', conn)
# Showing sale_date, product, and price from the sales table
pd.read_sql_query('''
SELECT sale_date, product, price
FROM sales
''', conn)
```

# 2. WHERE: Human Resources
Find employees in Chicago with more than 10 years of experience using WHERE and AND.

```python
# Load hr_employees.csv into the 'employees' table
pd.read_csv('hr_employees.csv').to_sql('employees', conn, if_exists='replace', index=False)
# Preview 5 rows from the employees table
pd.read_sql_query('''
SELECT * FROM employees LIMIT 5
''', conn)
# Showing employees in Chicago with more than 10 years of experience using WHERE and AND
pd.read_sql_query('''
SELECT name, department, city, salary, years_experience
FROM employees
WHERE city = 'Chicago'
AND years_experience > 10
''', conn)
```

# 3. ORDER BY: Real Estate
Display the three largest properties by square footage using ORDER BY and LIMIT.

```python
# Load realestate_listings.csv into the 'listings' table
pd.read_csv('realestate_listings.csv').to_sql('listings', conn, if_exists='replace', index=False)
# Preview 5 rows from the listings table
pd.read_sql_query('''
SELECT * FROM listings LIMIT 5
''', conn)
# Displaying the three largest properties by square footage using ORDER BY and LIMIT
pd.read_sql_query('''
SELECT neighborhood, property_type, list_price, sqft
FROM listings
ORDER BY sqft DESC
LIMIT 3
''', conn)
```

# 4. GROUP BY: Marketing
Calculate total weekly impressions using SUM() and GROUP BY.

```python
# Load marketing_campaigns.csv into the 'campaigns' table
pd.read_csv('marketing_campaigns.csv').to_sql('campaigns', conn, if_exists='replace', index=False)
# Preview 5 rows from the campaigns table
pd.read_sql_query('''
SELECT * FROM campaigns LIMIT 5
''', conn)
# Calculating total weekly impressions using SUM() and GROUP BY
pd.read_sql_query('''
SELECT week, SUM(impressions) AS total_impressions
FROM campaigns
GROUP BY week
ORDER BY week
''', conn)
```

# 5. HAVING: Restaurant
Find dates when total restaurant sales exceeded $400.

```python
# Load restaurant_orders.csv into the 'orders' table
pd.read_csv('restaurant_orders.csv').to_sql('orders', conn, if_exists='replace', index=False)
# Preview 5 rows from the orders table
pd.read_sql_query('''
SELECT * FROM orders LIMIT 5
''', conn)
# Finding order dates when total restaurant sales exceeded $400
pd.read_sql_query('''
SELECT order_date, ROUND(SUM(order_total), 2) AS total_sales
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

# 6. Aggregate Functions: Healthcare
Find the average patient age and longest waiting time for each doctor.

```python
# Load clinic_visits.csv into the 'visits' table
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists='replace', index=False)
# Preview 5 rows from the visits table
pd.read_sql_query('''
SELECT * FROM visits LIMIT 5
''', conn)
# Calculating the average patient age and longest waiting time for each doctor
pd.read_sql_query('''
SELECT doctor,
       ROUND(AVG(patient_age), 2) AS avg_patient_age,
       MAX(wait_minutes) AS max_wait_minutes
FROM visits
GROUP BY doctor
''', conn)
```

# 7. JOIN: Banking
Join customer and transaction tables to calculate total transaction amounts by branch.

```python
# Load both bank files into the 'customers' and 'transactions' tables
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists='replace', index=False)
# Joining customer and transaction tables to calculate total transaction amounts by branch
pd.read_sql_query('''
SELECT c.branch,
       ROUND(SUM(t.amount), 2) AS total_transaction_amount
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
